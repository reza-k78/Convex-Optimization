# PART B

```python
#n = 5
#N = 50  number of items
#S => set of similar pairs
#D => set of dissimilar pairs
#X = (N, n) feature vectors
P = cp.Variable((n,n), PSD=True)

# Loss function
L_S = cp.sum([(X[i]-X[j]).T @ P @ (X[i]-X[j]) for i, j in S])
L_D = cp.sum([cp.inv_pos((X[i]-X[j]).T @ P @ (X[i]-X[j])) for i, j in D])
L = L_S + L_D

objective = cp.Minimize(L)
#constraints = [P>>0] # and you have to enable symmetric=True in P (instead of PSD=True)
problem = cp.Problem(objective)
problem.solve()

print("status:", problem.status)
print("Loss function Value:", problem.value)
optimal_P = P.value
print("Optimal P:\n", optimal_P)
```

# PART C

```python
def distance(x, y):
  return (x - y).T @ optimal_P @ (x - y)

# Calculate training data misclassifications
num_misclassified_train = 0
for i, j in S:
  if distance(X[i], X[j]) > 1:
    num_misclassified_train += 1
for i, j in D:
  if distance(X[i], X[j]) <= 1:
    num_misclassified_train += 1

total_pairs_train = len(S) + len(D)
training_error_rate = num_misclassified_train / (total_pairs_train)

# Calculate test data misclassifications
num_misclassified_test = 0
for i, j in S_test:
  if distance(X[i], X[j]) > 1:
    num_misclassified_test += 1
for i, j in D_test:
  if distance(X[i], X[j]) <= 1:
    num_misclassified_test += 1

total_pairs_test = len(S_test) + len(D_test)
test_error_rate = num_misclassified_test / (total_pairs_test)

print("Training Error Rate:", training_error_rate)
print("Test Error Rate:", test_error_rate)
```
